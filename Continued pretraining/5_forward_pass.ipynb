```python
from transformers import AutoTokenizer
from transformers import AutoModelForCausalLM
import torch

model_name = "Qwen/Qwen2.5-0.5B"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
```

```python
text = "Python is a programming language"
inputs = tokenizer(
    text,
    return_tensors = 'pt'
)
print(inputs)
```

```python
print(inputs["input_ids"].shape)
#batch_size = x, sequence_lenght = y
```

```python
output = model(**inputs)
```

```python
print(output.keys())
```

```python
print(output['logits'].shape)
# x = batch size, y = input tokens, z = vocabulary size
```

```python
print(output.logits[0][0])
```

```python
predicted_token = output.logits[0][0].argmax()
print(predicted_token)
```

```python
print(
    tokenizer.decode(predicted_token)
)
```

```python
outputs = model(
    **inputs,
    labels=inputs["input_ids"]
)
```

```python
print(outputs.loss)
```

```python
print(inputs["input_ids"])

print(inputs["input_ids"].shape)

print(outputs.logits.shape)

print(outputs.loss)

print(outputs.logits[0][0][:10])

print(outputs.logits[0][0].argmax())

print(f"Decoded: {tokenizer.decode(outputs.logits[0][0].argmax())}")
```
